## 1. `with ... open as` pattern

```python
# Create a file called data.txt
with open("data.txt", "w") as file:
    file.write("Hello, this is my data.")
```

```python
# read the file data.txt in read only mode
with open('data.txt', 'r') as f:
    data = f.read()
```

```python
# write more data into the file
with open('data.txt', 'w') as f:
    data = 'some data to be written to the file'
    f.write(data)
```

## 2. Getting a Directory Listing

#### Directory Listings

```python
import os
#os.listdir() is legacy and returns a Python list containing the names of the files and subdirectories in directory
entries = os.listdir('/Users/Kelly/Documents/Documents/cs315') 
entries
```

```python
for entry in entries:
    print(entry) #cleaner output
```

#### Alternatively, use os.scandir() and pathlib.Path() in modern Python versions

```python
entries2 = os.scandir('/Users/Kelly/Documents/Documents/cs315')
entries2
```

```python
# loop through using with()
with os.scandir('/Users/Kelly/Documents/Documents/cs315') as entries2:
    for entry2 in entries2:
        print(entry2.name)
```

```python
# using pathlib module
from pathlib import Path

entries = Path('/Users/Kelly/Documents/Documents/cs315')
for entry in entries.iterdir():
    print(entry.name)
```

**Using pathlib.Path() or os.scandir() instead of os.listdir() is the preferred way of getting a directory listing**

#### Listing All Files in Directory

```python
# List all files in a directory using os.listdir
basepath = '/Users/Kelly/Documents/Documents/cs315/'
for entry in os.listdir(basepath):
    if os.path.isfile(os.path.join(basepath, entry)):
        print(entry)
```

```python
# using os.scandir()
with os.scandir(basepath) as entries:
    for entry in entries:
        if entry.is_file():
            print(entry.name)
```

Using os.scandir() has the advantage of looking cleaner and being easier to understand than using os.listdir()

```python
# List all files in directory using pathlib
from pathlib import Path

basepath = Path('/Users/Kelly/Documents/Documents/cs315/')

files_in_basepath = (entry for entry in basepath.iterdir() if entry.is_file())
for item in files_in_basepath:
    print(item.name)
```

#### Listing Subdirectories

```python
# List all subdirectories using os.listdir
basepath = '/Users/Kelly/Documents/Documents/cs315/'
for entry in os.listdir(basepath):
    if os.path.isdir(os.path.join(basepath, entry)):
        print(entry)
```

```python
# list subdirectories using os.scandir()
with os.scandir(basepath) as entries:
    for entry in entries:
        if entry.is_dir():
            print(entry.name)
```

```python
# using pathlib.Path()
basepath = Path('/Users/Kelly/Documents/Documents/cs315/')
for entry in basepath.iterdir():
    if entry.is_dir():
        print(entry.name)
```

## 3. Getting File Attributes

```python
import os
with os.scandir('/Users/Kelly/Documents/Documents/cs315/') as dir_contents:
     for entry in dir_contents:
         info = entry.stat()
         print(info.st_mtime)

#returns a ScandirIterator object, where each entry has a .stat() method that retrieves information about the file or directory it points to
```

## 4. Making Directories

os.mkdir(): Creates a single subdirectory\
pathlib.Path.mkdir(): Creates single or multiple directories\
os.makedirs(): Creates multiple directories, including intermediate directories

```python
import os

os.mkdir('example_directory/')
```

```python
from pathlib import Path

p = Path('example_directory')
try:
    p.mkdir()
except FileExistsError as exc:
    print(exc)
```

```python
p = Path('example_directory')
p.mkdir(exist_ok=True)
```

#### Creating Multiple Directories

os.makedirs() can make not only individual directories, but also directory trees

```python
os.makedirs('2026/09/08')
```

```python
import pathlib

p = pathlib.Path('2018/10/05')
p.mkdir(parents=True)
```
